```python
import requests, os, re
from bs4 import BeautifulSoup
from dotenv import load_dotenv
load_dotenv()

HEADERS = {"User-Agent": os.getenv("SEC_USER_AGENT")}
url     = "https://www.sec.gov/Archives/edgar/data/1326801/000162828025047240/meta-20250930.htm"
html    = requests.get(url, headers=HEADERS).text

soup = BeautifulSoup(html, "lxml")
for tag in soup(["script","style"]): tag.decompose()
for tag in soup.find_all(["ix:header","header"]): tag.decompose()
for tag in soup.find_all("ix:hidden"): tag.decompose()

text = soup.get_text(separator="\n")
lines = [l.strip() for l in text.splitlines() if l.strip()]

# Find Item/Part markers
markers = [l for l in lines if re.search(
    r'\b(?:ITEM|Item|PART|Part)\s+\d+|'
    r'\b(?:ITEM|Item|PART|Part)\s+[IVX]+', l
) and len(l) < 150]

print(f"Total marker candidates: {len(markers)}")
for m in markers[:30]:
    print(f"  {repr(m)}")
```
